```python
#Reverse a number using a loop.
n=int(input("Enter the Number: "))
rev=0;
while n > 0:
  d=n%10
  rev=rev*10 +d
  n=n//10
print ("Reversed number: ",rev)
```

```python
#Check whether a number is prime.
n=int(input("Enter the Number: "))
if n<=1:
  print("Not A prime Number")
else:
  for i in range (2,int (n**0.5)+1):
    if n%i==0:
      print("Not A Prime Number")
      break
  else:
      print("Prime Number")

```

```python
#Write a function to check if a number is palindrome
n=int(input("Enter the Number: "))
o=n
rev=0
while n!=0:
  d=n%10
  rev=rev*10 + d
  n=n//10
if rev==o:
  print("Paindrome Number")
else:
  print("Not a Paindrome Number")
```

```python
#Write a function to calculate sum of digits of a number.
def sum_of_digits(number):
    sum = 0
    while number > 0:
        digit = number % 10
        sum += digit
        number //= 10
    return sum
n=int(input("Enter the Number: "))
print("Sum = ",sum_of_digits(n))
```
